```python
import pandas as pd

df = pd.read_csv('heart_disease_uci.csv')
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.groupby('sex')['num'].mean()
```

```python
import matplotlib.pyplot as plt

df.groupby('sex')['num'].mean().plot(kind='bar')
plt.title('Average heart disease severity by sex')
plt.ylabel('Average severity score (0-4)')
plt.show()
```

```python
df.groupby('age')['num'].mean()
```

```python
import matplotlib.pyplot as plt

df.groupby('age')['num'].mean().plot(kind='bar')
plt.title('Average heart disease severity by age')
plt.ylabel('Average severity score (0-4)')
plt.show()
```

```python
df.groupby('sex')['chol'].mean().plot(kind='bar')
plt.title('Average cholesterol by sex')
plt.ylabel('Cholesterol')
plt.show()
```

```python
df.groupby('age')['chol'].mean().plot(kind='bar')
plt.title('Average cholesterol by age')
plt.ylabel('Cholesterol')
plt.show()
```
